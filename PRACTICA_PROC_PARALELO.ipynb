```python
import time
import multiprocessing
import os
#----------------------------------------------------------------------------
# CREAMOS LA FUNCION PARA EL PROCESAMIENTO
#----------------------------------------------------------------------------
def calculo(n):
  res = 0
  for i in range(20000000):
    res=res + ((i%5)*(i%7))
  return res

if __name__== "__main__":

#----------------------------------------------------------------------------
#consultamos nuemero de CPU()
#----------------------------------------------------------------------------

  n_cpu=os.cpu_count()
  print("La cantidad de CPU asignada es",n_cpu)

#----------------------------------------------------------------------------
#Ingresamos el numero de repeticiones para la funcion
#----------------------------------------------------------------------------

  rep=int(input("Ingresar numero de repeticiones"))

#----------------------------------------------------------------------------
#INICIAMOS PROCESAMIENTO SECUENCIAL
#----------------------------------------------------------------------------
  print("SE INICIO PROCESO SECUENCIAL")
  inicio_tps = time.time()

  for i in range(rep):
    calculo(i)

  fin_tps=time.time()

  tiempo_ps=round(fin_tps - inicio_tps,4)

  print("El tiempo que se demoro en procesar de forma SECUENCIAL es:",tiempo_ps)

#----------------------------------------------------------------------------
# INICAMOS PROCESAMIENTO PARALELO
#----------------------------------------------------------------------------

print("SE INICIO EL PROCESAMIENTO PARALELO")

cantidad_cpu= 2

inicio_tpp=time.time()

with multiprocessing.Pool(processes=cantidad_cpu) as pool:
  pool.map(calculo,range(rep))

fin_tpp=time.time()

tiempo_pp=round(fin_tpp - inicio_tpp,4)
print("El tiempo que se demoro en procesar de forma PARALELA es:",tiempo_pp)

#----------------------------------------------------------------------------
# CALCULO DE PARAMETROS
#----------------------------------------------------------------------------

speedup= tiempo_ps/tiempo_pp

eficiencia= (speedup/cantidad_cpu)*100

mejora= ((tiempo_ps-tiempo_pp)/tiempo_ps)*100


#----------------------------------------------------------------------------
#VISUALIZACION DE RESULTADOS
#----------------------------------------------------------------------------

print("********RESULTADOS*********")
print("Tiempo Prosesamiento Secuencial: ",tiempo_ps)
print("Tiempo Prosesamiento Paralelo: ",tiempo_pp)
print("Procesadores utilizados : ",cantidad_cpu)
print("Parametro Speedup : ",speedup)
print("Parametro Eficiencia : ",eficiencia)
print("Parametro mejora : ",mejora)
print("***************************")
```
